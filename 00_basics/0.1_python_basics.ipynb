# Python基础语法


### 变量

- 变量是将数据在内存中存储后给定的一个名称，这个名称就是变量。
- 变量名是没有类型的，类型只存在于对象中，变量只是引用对象。
- Python 为动态解释型语言，在数值操作时，类型是在运行过程中自动决定的，无需事先声明。
- 变量命名规则：
   - 变量名只能由数字、字母和下划线(_)组成
   - 变量名不能以数字开头
   - 变量名不能是 Python 关键字

### 对象

- 对象是有类型的。
- 对象是分配的一块内存空间，用来表示它的值。
- Python 对象的三要素：
  - `id` 对象的唯一标识符
  - `type` 对象的类型
  - `value` 对象的值

完整语法格式：


`value` ——— 要打印的东西（数量无限制）

## 2. Python 的数据类型

### 2.1 Number 数值类型

Python 的数字类型(Number)一共有四种:

| 类型 | 关键字 | 说明 | 例子 |
|---|---|---|---|
| 整数 | `int` | 不带小数点,任意精度 | `42`、`-7`、`0` |

````python
# 做除法后的结果类型为浮点数
a = 11
b = 3
c = a/b # <class 'float'>
print(c)

# 整形的整除
d = a//b # 会将小数部分丢弃
print(d)

# 四舍五入
print(round(3.66666,2))

# 幂操作
e = a**2
print(e)

# 只保留整数部分
f = int(c)
print(f)
````

### 2.2 String 字符串类型

字符串用单引号 `'...'` 或双引号 `"..."` 包起来。它是**不可变(immutable)**的:一旦创建,里面的字符就不能改。

### 索引和切片

- **索引**:`s[0]` 是第一个字符,`s[-1]` 是最后一个(负数从右往左数)
- **切片**:`s[起点:终点:步长]`,**左闭右开**(含起点、不含终点)
  - `s[0:3]` → 第 0、1、2 个字符
  - `s[::2]` → 每隔一个取
  - `s[::-1]` → 倒序
- 当我们定义 `a = "Python"` 时,计算机的执行顺序:
   - 先在内存中**创建字符串对象** `"Python"`(分配内存,自带 `id` / `type` / `value`)
   - 再创建变量名 `a`
   - 最后让 `a` **引用(指向)** 这个对象，即把对象的内存地址赋给a

  所以 `a` 不是"容器",只是指向对象的一个”指针“;两个变量可以指向同一个对象:

  ```python
  a = "Python"
  b = a                  # b 和 a 指向同一个对象
  print(id(a) == id(b))  # True,地址相同
  ```

### 常用方法

| 方法 | 作用 | 例子 | 结果 |
|---|---|---|---|
| `len(s)` | 长度 | `len("Miku")` | `4` |
| `s.upper()` | 全大写 | `"miku".upper()` | `"MIKU"` |
| `s.lower()` | 全小写 | `"MIKU".lower()` | `"miku"` |
| `s.strip()` | 去两端空白 | `" hi ".strip()` | `"hi"` |
| `s.replace(a, b)` | 替换 | `"a,b".replace(",", "-")` | `"a-b"` |
| `s.split(",")` | 按逗号拆分 | `"a,b".split(",")` | `["a", "b"]` |
| `s.find("b")` | 查找位置(找不到返回 -1) | `"abc".find("b")` | `1` |
| `"-".join(list)` | 用分隔符连接列表 | `"-".join(["a","b"])` | `"a-b"` |
| `"i" in s` | 判断是否包含 | `"i" in "Miku"` | `True` |

````python
s = "Hello, Miku"

# 索引和切片
print(s[0])        # H
print(s[-1])       # u
print(s[0:5])      # Hello
print(s[::2])      # Hlo iu (每隔一个)
print(s[::-1])     # ukiM ,olleH (倒序)

# 常用方法
print(len(s))                    # 11
print(s.upper())                 # HELLO, MIKU
print("  hi  ".strip())          # hi
print(s.replace("Miku", "world"))  # Hello, world
print(s.split(","))              # ['Hello', ' Miku']
print("i" in s)                  # True

# 字符串是不可变的:下面这行会报错,可以取消注释试试
# s[0] = "h"   # TypeError: 'str' object does not support item assignment
````

### 2.3 List 列表

列表用方括号 `[...]` 包起来,元素可以是任意类型。

和字符串**最大的不同:列表是可变(mutable)的**——能改元素、增删元素。索引和切片规则与字符串**完全一样**。

##### 常用方法

| 方法 | 作用 |
|---|---|
| `lst.append(x)` | 末尾添加一个元素 |
| `lst.insert(i, x)` | 在位置 i 插入 |
| `lst.pop()` | 删除并返回末尾元素 |
| `lst.remove(x)` | 删除第一个值为 x 的元素 |
| `lst.sort()` | 原地排序(改原列表) |
| `sorted(lst)` | 返回新列表(不改原列表) |
| `lst.reverse()` | 原地反转 |
| `lst.index(x)` | 查找第一次出现的位置 |
| `x in lst` | 判断是否包含 |
| `len(lst)` | 长度 |

````python
lst = [1, 2, 3, 4, 5]
print(lst[0])      # 1
print(lst[-1])     # 5
print(lst[1:3])    # [2, 3] 左闭右开

# 可变:直接改元素
lst[0] = 99
print(lst)         # [99, 2, 3, 4, 5]

# 常用方法
nums = [3, 1, 2]
nums.append(4)
print(nums)        # [3, 1, 2, 4]
nums.sort()
print(nums)        # [1, 2, 3, 4]
print(nums.pop())  # 4 (弹出末尾)
````

##### 易错点

列表是可变对象,`b = a` 只是让 `b` 也指向**同一个列表**:

```python
a = [1, 2, 3]
b = a            # 同一个对象
b.append(4)
print(a)         # [1, 2, 3, 4]  ← a 也跟着变了!
```

想得到**独立副本**,用 `b = a.copy()` 或 `b = a[:]`。

##### 列表推导式

语法:`[表达式 for 变量 in 可迭代对象]`,可加 `if` 过滤:

```python
[x**2 for x in range(5)]              # [0, 1, 4, 9, 16]
[x for x in range(10) if x % 2 == 0]  # [0, 2, 4, 6, 8]
```

````python
# 引用陷阱
a = [1, 2, 3]
b = a              # 同一个对象
b.append(99)
print(a)           # [1, 2, 3, 99]  ← a 也变了!

c = a.copy()       # 独立副本
c.append(100)
print(a, c)        # [1, 2, 3, 99] [1, 2, 3, 99, 100]

# 列表推导式
print([x**2 for x in range(5)])              # [0, 1, 4, 9, 16]
print([x for x in range(10) if x % 2 == 0])  # [0, 2, 4, 6, 8]
````

#### 2.4 Tuple 元组

元组用圆括号 `(...)` 定义,和列表很像,但有一个关键区别:**元组是不可变(immutable)的**——创建后不能增删改元素。

| | 列表 list | 元组 tuple |
|---|---|---|
| 符号 | `[1, 2, 3]` | `(1, 2, 3)` |
| 可变性 | 可变 | **不可变** |
| 用途 | 需要增删改的数据 | 不想被改的数据(坐标、配置) |

**几个要点**

- 索引和切片规则与列表 / 字符串**完全一样**
- **单元素元组要加逗号**:`(1,)` 才是元组,`(1)` 只是数字 `1`
- **解包**:`x, y = (1, 2)` 一行给多个变量赋值
- 元组可以作为字典的 key,列表不行(因为元组不可变、可哈希)

````python
t = (1, 2, 3, 4, 5)
print(type(t))       # <class 'tuple'>
print(t[0])          # 1
print(t[1:3])        # (2, 3)

# 不可变:下面这行会报错
# t[0] = 99   # TypeError: 'tuple' object does not support item assignment

# 单元素元组要加逗号
print(type((1,)))    # <class 'tuple'>
print(type((1)))     # <class 'int'>  ← 这是数字 1,不是元组!

# 解包
x, y, z = (10, 20, 30)
print(x, y, z)       # 10 20 30

# 交换两个变量(利用解包)
a, b = 1, 2
a, b = b, a
print(a, b)          # 2 1
````

#### 2.5 Set 集合

集合用花括号 `{...}` 定义,有两个特点:**元素无序、不重复**。

**主要用途**

- **去重**:自动去掉重复元素
- **集合运算**:交集、并集、差集(就是数学里的集合)

**常用操作**

| 操作 | 含义 |
|---|---|
| `set([...])` | 从列表重建集合(去重) |
| `s.add(x)` | 添加元素 |
| `s.remove(x)` | 删除元素 |
| `a & b` | 交集(两边都有) |
| `a \| b` | 并集(任一边有) |
| `a - b` | 差集(在 a 不在 b) |
| `x in s` | 判断是否包含 |

> 集合不能索引(无序),也不能放可变元素(比如列表)。

````python
# 去重
nums = [1, 2, 2, 3, 3, 3]
print(set(nums))       # {1, 2, 3}

# 集合运算
a = {1, 2, 3}
b = {2, 3, 4}
print(a & b)           # {2, 3} 交集
print(a | b)           # {1, 2, 3, 4} 并集
print(a - b)           # {1} 差集

# 添加 / 删除
s = {1, 2}
s.add(3)
print(s)               # {1, 2, 3}
````

#### 2.6 Dict 字典

字典用 `{key: value}` 定义,存的是**键值对**,像查表一样通过 key 找 value。

**要点**

- **键 key 必须是不可变的**(字符串、数字、元组),值 value 可以是任意类型
- 取值:`d[key]`;但 key 不存在会报错,用 `d.get(key, 默认值)` 更安全
- 增改:`d[key] = value`(键存在就改,不存在就加)

**常用方法**

| 方法 | 作用 |
|---|---|
| `d[key]` | 取值(键不存在会报错) |
| `d.get(key, 默认)` | 安全取值(不存在返回默认值) |
| `d[key] = v` | 添加或修改 |
| `del d[key]` | 删除 |
| `key in d` | 判断键是否存在 |
| `d.keys()` | 所有键 |
| `d.values()` | 所有值 |
| `d.items()` | 所有键值对(遍历用) |

````python
d = {"name": "Miku", "age": 16}
print(d["name"])              # Miku
print(d.get("height", "unknown"))  # unknown (键不存在,返回默认)

# 添加 / 修改
d["age"] = 16                 # 修改
d["city"] = "Tokyo"           # 添加
print(d)                      # {'name': 'Miku', 'age': 16, 'city': 'Tokyo'}

# 遍历键值对
for k, v in d.items():
    print(k, "=", v)

# 判断键是否存在
print("age" in d)             # True
````

## 3. Python 三大程序结构

### 3.1 顺序结构

程序默认按**从上到下的顺序,一行一行执行**,这就是顺序结构——最基本、最自然的执行方式。

每条语句按书写顺序依次执行,执行完上面才轮到下面。

````python
print("Step 1")
print("Step 2")
print("Step 3")
# 依次输出:Step 1 → Step 2 → Step 3
````

### 3.2 分支结构

分支结构让程序根据**条件**走不同的路,用 `if` / `elif` / `else` 实现。

**基本语法**

```python
if 条件:
    满足条件时执行的代码
elif 另一个条件:
    满足另一个条件时执行的代码
else:
    都不满足时执行的代码
```

**三个要点**

- 条件后面要写**冒号 `:`**
- 冒号下面要**缩进 4 个空格**(Python 靠缩进区分代码块,不是大括号!)
- `elif`、`else` 都可以省略

**比较运算符**:`==` 等于、`!=` 不等于、`>` `<` `>=` `<=`

**逻辑运算符**:`and`(且)、`or`(或)、`not`(非)

````python
score = 85

if score >= 90:
    print("Excellent")
elif score >= 60:
    print("Pass")
else:
    print("Fail")

# 逻辑运算:and 表示"两个条件都满足"
age = 20
if age >= 18 and age < 60:
    print("adult")
````

### 3.3 循环结构

循环结构让程序重复执行一段代码,有两种:

### `for` 循环 —— 遍历一个序列

```python
for 变量 in 可迭代对象:
    循环体
```

`range(n)` 生成 `0, 1, 2, ..., n-1`(不含 n)。可以遍历列表、字符串、字典等。

### `while` 循环 —— 条件成立就一直跑

```python
while 条件:
    循环体
```

**要点**

- 和 if 一样:冒号 + 缩进
- `break` 直接跳出**整个循环**
- `continue` 跳过**本次**,进入下一次
- while 要小心**死循环**(条件一直为真,程序停不下来)

````python
# for + range:打印 0 到 4
for i in range(5):
    print(i, end=" ")   # 0 1 2 3 4
print()

# 遍历列表
for name in ["Miku", "Rin", "Len"]:
    print("Hello,", name)

# while:累加到 100
total = 0
i = 1
while i <= 100:
    total += i
    i += 1
print("Sum from 1 to 100:", total)   # 5050

# break 和 continue
for i in range(10):
    if i == 3:
        continue   # 跳过 3
    if i == 8:
        break      # 到 8 就停
    print(i, end=" ")
# 输出:0 1 2 4 5 6 7
````

````python
# 打印九九乘法表
## while 循环
i = 1
while(i < 10):
    j = 1;
    while(j <= i):
        print("%d*%d=%d" % (i,j,i*j),end = '\t')
        j = j + 1
    i = i + 1
    print("\n")

## for 循环
for i in range(1,10):
    for j in range(1,i+1):
        print(f"{i}*{j}={i*j}", end = '\t')
    print("\n")
````

## 4. 函数

前面写的代码都在用别人造好的轮子(`print`、`len`……),这一节学**自己造轮子**。

函数就是**把一段代码打包、起个名字**,以后想用直接喊名字,不用重复写。

好处:

- **好维护** —— 要改只改函数里那一处

本节先简单看看 Python 自带的内置函数,再学怎么定义自己的函数。


### 4.1 内置函数

**内置函数**是 Python 自带、装好就能直接用的函数,不用自己写,也不用 `import`。

| 函数 | 作用 | 例子 | 结果 |
|---|---|---|---|
| `len(x)` | 长度 | `len("Miku")` | `4` |
| `type(x)` | 类型 | `type(3.14)` | `<class 'float'>` |
| `max(x)` / `min(x)` | 最大 / 最小 | `max([3, 1, 2])` | `3` |
| `sum(x)` | 求和 | `sum([1, 2, 3])` | `6` |
| `abs(x)` | 绝对值 | `abs(-5)` | `5` |
| `round(x, n)` | 四舍五入 | `round(3.14159, 2)` | `3.14` |
| `sorted(x)` | 排序(返回新列表) | `sorted([3, 1, 2])` | `[1, 2, 3]` |

> 第 1 节的 `print()`、`input()` 也是内置函数。完整列表见官方文档:<https://docs.python.org/zh-cn/3/library/functions.html>


````python
# 内置函数直接调用,不需要 import
nums = [3, 1, 4, 1, 5]

print(len(nums))          # 5    长度
print(max(nums))          # 5    最大值
print(min(nums))          # 1    最小值
print(sum(nums))          # 14   求和
print(sorted(nums))       # [1, 1, 3, 4, 5]  排序(返回新列表)
print(abs(-3.14))         # 3.14 绝对值
print(round(3.14159, 2))  # 3.14 保留两位小数
````

### 4.2 自定义函数

自定义函数用 `def` 关键字定义:

```python
def 函数名(参数列表):
    函数体
    return 返回值
```

**要点**

- `def` 开头,函数名后面跟**圆括号**和**冒号**
- 函数体要**缩进 4 个空格**(和 if / for 一样靠缩进区分代码块)
- `return` 把结果交给调用者;不写 `return` 时,函数默认返回 `None`
- **定义**函数不会执行里面的代码,只有**调用**(写 `函数名()`)时才执行


````python
# 定义:无参数、无返回值
def say_hello():
    print("Hello, world!")

say_hello()
say_hello()

# 定义:带参数、带返回值
def add(a, b):
    return a + b

result = add(3, 5)
print(result)        # 8
````

调用时传给函数的值叫**参数**。常见的三种写法:

| 类型 | 写法 | 说明 |
|---|---|---|
| 位置参数 | `def f(a, b)` | 按顺序传:`f(1, 2)` |
| 默认参数 | `def f(a, b=10)` | 不传就用默认值;默认参数要放在后面 |
| 关键字参数 | 调用时 `f(b=2, a=1)` | 按名字传,顺序可以打乱 |

```python
def power(base, exp=2):     # exp 有默认值 2
    return base ** exp

power(3)          # 9   → exp 用默认值
power(3, 3)       # 27  → 按位置传
power(exp=3, base=2)  # 8  → 按关键字传
```


````python
def power(base, exp=2):      # exp 是默认参数,默认平方
    return base ** exp

print(power(3))              # 9   → 用默认 exp=2
print(power(3, 3))           # 27  → 按位置传 exp=3
print(power(exp=3, base=2))  # 8   → 按关键字传,顺序随意

def greet(name, greeting="Hello"):
    print(greeting + ", " + name + "!")

greet("Miku")
greet("Miku", "Hi")
````

`return` 决定函数**交出去的结果**:

- 没写 `return`(或只写 `return`)→ 返回 `None`
- 想返回**多个值**,用逗号隔开 —— 实际返回的是一个**元组**
- 函数执行到 `return` 就**立刻结束**,后面的代码不会跑

```python
def min_max(nums):
    return min(nums), max(nums)

lo, hi = min_max([3, 1, 4, 1, 5])   # 解包成两个变量
```


````python
# 没有 return → 返回 None
def no_return():
    x = 1 + 1

print(no_return())           # None

# 返回多个值:实际是一个元组
def min_max(nums):
    return min(nums), max(nums)

print(min_max([3, 1, 4, 1, 5]))     # (1, 5)
lo, hi = min_max([3, 1, 4, 1, 5])   # 解包
print(lo, hi)                       # 1 5

# return 之后的代码不会执行
def sign(n):
    if n < 0:
        return "negative"
    return "non-negative"

print(sign(-1))    # negative
print(sign(5))     # non-negative
````

函数里定义的变量是**局部变量**,只在函数内部存在;函数外定义的叫**全局变量**。

```python
x = 10          # 全局

def f():
    x = 20      # 这是新的局部变量,和外面的 x 无关
    print(x)

f()             # 20
print(x)        # 10  ← 外面的 x 没变
```

**要点**:函数内部**读**全局变量没问题,但想**改**它需要先用 `global` 声明。初学阶段尽量别用 `global` —— 数据靠**参数传进去、`return` 传出来**更清晰。


````python
x = 10                 # 全局变量

def f():
    x = 20             # 局部变量,和外面的 x 无关
    print("inside:", x)

f()                    # inside: 20
print("outside:", x)   # outside: 10
````

## 5. Python 面向对象编程

前面我们都是"面向过程"地写代码:一行行按步骤执行。**面向对象(OOP)** 换一种思路——把**数据**和**操作数据的方法**打包成一个整体,这个整体就是**对象**。

这一节要弄明白两组词:

| 概念 | 说明 | 类比 |
|---|---|---|
| **类 class** | 模板 / 图纸 | 造车的图纸 |
| **对象 object** | 按模板造出来的具体东西 | 图纸造出的一辆车 |

先会定义类、创建对象,再学继承和魔法方法。


### 5.1 类与对象

```python
class 类名:              # 类名习惯首字母大写
    def 方法名(self):     # 方法就是"写在类里的函数"
        ...
```

- 用 `class` 关键字定义类,类名一般**首字母大写**(`Dog`、`Person`)
- 用 `类名()` **创建对象**,这个过程叫**实例化**
- 对象 = **属性**(数据)+ **方法**(行为)

```python
class Dog:
    def __init__(self, name, age):   # 创建对象时自动调用
        self.name = name             # 属性
        self.age = age

    def bark(self):                  # 方法
        print(self.name + " says woof!")

d = Dog("Buddy", 3)     # 实例化:造出一个具体的狗
print(d.name, d.age)   # Buddy 3
d.bark()               # Buddy says woof!
```


````python
class Dog:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def bark(self):
        print(self.name + " says woof!")

d = Dog("Buddy", 3)       # 实例化:按 Dog 这个模板造一个对象
print(type(d))           # <class '__main__.Dog'>
print(d.name, d.age)     # Buddy 3
d.bark()                 # Buddy says woof!
````

### 5.2 类中的属性和方法

类里面装两类东西:**属性**(数据)和**方法**(行为)。而属性又分两种:

| 类型 | 定义位置 | 特点 |
|---|---|---|
| **实例属性** | `__init__` 里 `self.xxx = ...` | 每个对象各有一份 |
| **类属性** | 直接写在 class 里(方法外) | 所有对象**共享同一份** |

```python
class Dog:
    species = "canine"      # 类属性:所有狗共享

    def __init__(self, name):
        self.name = name    # 实例属性:每只狗自己的名字
```

- 读类属性:`Dog.species` 或 `d.species` 都行
- 改类属性:要用 `Dog.species = ...`;用 `d.species = ...` 反而会**新建一个实例属性**,不是改类属性(看下面的例子)


````python
class Dog:
    species = "canine"        # 类属性:所有狗共享

    def __init__(self, name):
        self.name = name      # 实例属性:每只狗自己的

a = Dog("Buddy")
b = Dog("Bella")

# 实例属性:各自独立
print(a.name, b.name)        # Buddy Bella

# 类属性:共享同一份
print(a.species, b.species)  # canine canine

# 改类属性 → 所有对象都跟着变
Dog.species = "mammal"
print(a.species, b.species)  # mammal mammal

# 注意:用实例改,会新建一个实例属性,而不是改类属性
a.species = "bird"
print(a.species, b.species)  # bird mammal  ← 只有 a 自己变了
````

方法(写在类里的函数)有三种,靠**装饰器**区分:

| 类型 | 写法 | 第一个参数 | 调用 |
|---|---|---|---|
| **实例方法** | `def m(self)` | 对象自己 | `obj.m()` |
| **类方法** | `@classmethod def m(cls)` | 类本身 | `类.m()` 或 `obj.m()` |
| **静态方法** | `@staticmethod def m()` | 无 | `类.m()` |

- **实例方法**最常用,前面的 `bark` 就是,必须通过对象调用
- **类方法**第一个参数是 `cls`(代表类本身),常用在"通过类做点什么、不用造对象"的场景
- **静态方法**就是普通函数,只是逻辑上和类相关,既不要 `self` 也不要 `cls`


````python
class Dog:
    species = "canine"

    def __init__(self, name):
        self.name = name

    def bark(self):                      # 实例方法:需要 self
        print(self.name + " says woof!")

    @classmethod
    def species_name(cls):               # 类方法:第一个参数是 cls(类本身)
        return cls.species

    @staticmethod
    def is_dog(name):                    # 静态方法:不需要 self / cls
        return name in ["Buddy", "Bella", "Milo"]

print(Dog.species_name())        # canine
print(Dog.is_dog("Buddy"))       # True
print(Dog.is_dog("Kitty"))       # False
````

### 5.3 `self` 和 `__init__`

这两个是初学最容易绕晕的地方:

- **`__init__`** —— 构造方法。创建对象时 **自动调用**,用来初始化属性,不用手动调。
- **`self`** —— 指**当前这个对象自己**,是方法的第一个参数。调用时不用传,Python 会自动补上。

```python
def __init__(self, name):
    self.name = name     # 把参数存到这个对象身上
```

所以 `Dog("Buddy", 3)` 其实做了两件事:

1. 造一个空的 `Dog` 对象
2. 自动把它作为 `self`,连同 `"Buddy", 3` 一起传给 `__init__`

> 记住:`self.xxx = xxx` 是**写属性**;`self.xxx` 是**读属性**。每个对象都有自己独立的一份属性,互相不影响。


````python
class Dog:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def bark(self):
        print(self.name + " says woof!")

a = Dog("Buddy", 3)
b = Dog("Bella", 2)

# 每个对象有自己独立的一份属性
print(a.name, a.age)   # Buddy 3
print(b.name, b.age)   # Bella 2

# 改一个对象的属性,不影响另一个
a.age = 4
print(a.age, b.age)    # 4 2

a.bark()               # Buddy says woof!
b.bark()               # Bella says woof!
````

### 5.4 继承

**继承**让一个类"白拿"父类的属性和方法,再自己添加新的,少写重复代码。

```python
class 子类(父类):
    ...
```

- `class Cat(Animal):` 表示 `Cat` 继承 `Animal`,自动拥有 `Animal` 的属性和方法
- 子类可以**重写(override)** 父类的方法:定义一个**同名**方法即可覆盖
- `super()` 用来调用父类的方法(常见于先跑父类的 `__init__`,再补充自己的)

```python
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        print("...")

class Cat(Animal):        # 继承 Animal
    def speak(self):      # 重写父类的方法
        print(self.name + " says meow")

class Dog(Animal):
    pass                  # 不写 speak,直接用父类的

Cat("Luna").speak()    # Luna says meow
Dog("Buddy").speak()  # ...   ← 用的父类方法
```


````python
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        print(self.name + " says ...")

class Cat(Animal):                 # 继承 Animal
    def speak(self):               # 重写父类方法
        print(self.name + " says meow")

class Dog(Animal):                 # 什么都不写,直接用父类的
    pass

Cat("Luna").speak()                 # Luna says meow
Dog("Buddy").speak()                # Buddy says ...

# super() 先调用父类的 __init__,再补充自己的属性
class Puppy(Animal):
    def __init__(self, name, age):
        super().__init__(name)     # 交给父类存 name
        self.age = age             # 自己再存 age

p = Puppy("Milo", 1)
print(p.name, p.age)               # Milo 1
````

### 5.5 魔法方法

名字以 `__` 开头、以 `__` 结尾的方法叫**魔法方法**(dunder method):不用手动调用,在特定时机**自动触发**。

| 方法 | 触发时机 |
|---|---|
| `__init__(self)` | 创建对象时 |
| `__str__(self)` | `print(对象)` 时 |
| `__len__(self)` | `len(对象)` 时 |

最常用的是 `__str__`:它决定对象被 `print` 时长什么样。不写的话,打印出来是一串看不懂的地址。

```python
class Dog:
    def __init__(self, name):
        self.name = name

    def __str__(self):                       # print(obj) 时显示的内容
        return "Dog(name=" + self.name + ")"

print(Dog("Buddy"))   # Dog(name=Buddy)
```


````python
class Dog:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):                    # print(obj) 时显示的内容
        return "Dog(name=" + self.name + ", age=" + str(self.age) + ")"

d = Dog("Buddy", 3)
print(d)              # Dog(name=Buddy, age=3)

# 没有 __str__ 时,打印的是一串内存地址,比如 <__main__.Dog object at 0x10a...>
class Plain:
    pass

print(Plain())        # <__main__.Plain object at 0x...>
````

### 5.6 这和机器学习有什么关系?

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()      # LinearRegression 是类;model 是实例对象
model.fit(X, y)                 # fit 是方法:训练

```

所以看到 `model.fit(...)`、`model.predict(...)` 不要发懵 —— 就是**创建对象、调用它的方法**而已。
