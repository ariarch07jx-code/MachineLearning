# Python基础语法


### 变量

- 变量是将数据在内存中存储后给定的一个名称，这个名称就是变量。
- 变量名是没有类型的，类型只存在于对象中，变量只是引用对象。
- Python 为动态解释型语言，在数值操作时，类型是在运行过程中自动决定的，无需事先声明。
- 变量命名规则：
   - 变量名只能由数字、字母和下划线(_)组成
   - 变量名不能以数字开头
   - 变量名不能是 Python 关键字

### 对象

- 对象是有类型的。
- 对象是分配的一块内存空间，用来表示它的值。
- Python 对象的三要素：
  - `id` 对象的唯一标识符
  - `type` 对象的类型
  - `value` 对象的值

 ### `print()` 输出函数

完整语法格式：

```python
print(*value, sep=" ", end="\n", file=None, flush=False)
```

`value` ——— 要打印的东西（数量无限制）

## 2. Python 的数据类型

### 2.1 Number 数值类型

Python 的数字类型(Number)一共有四种:

| 类型 | 关键字 | 说明 | 例子 |
|---|---|---|---|
| 整数 | `int` | 不带小数点,任意精度 | `42`、`-7`、`0` |
| 浮点数 | `float` | 带小数点的数 | `3.14`、`2.0`、`1e-3` |
| 复数 | `complex` | 实部 + 虚部,虚部用 `j` | `3+4j` |
| 布尔 | `bool` | 只有 `True` / `False` | `True`、`False` |

> 布尔 `bool` 本质是 `int` 的子类:`True` 等于 `1`,`False` 等于 `0`。

````python
# 用 type() 查看每个值的类型
print(type(42))       # <class 'int'>
print(type(3.14))     # <class 'float'>
print(type(3+4j))     # <class 'complex'>
print(type(True))     # <class 'bool'>

# 复数:实部 .real 和虚部 .imag
c = 3 + 4j
print(c.real, c.imag)   # 3.0 4.0

# 布尔本质是整数:True == 1,False == 0
print(True + 1)         # 2
````

````python
# 做除法后的结果类型为浮点数
a = 11
b = 3
c = a/b # <class 'float'>
print(c)

# 整形的整除
d = a//b # 会将小数部分丢弃
print(d)

# 四舍五入
print(round(3.66666,2))

# 幂操作
e = a**2
print(e)

# 只保留整数部分
f = int(c)
print(f)
````

### 2.2 String 字符串类型

字符串用单引号 `'...'` 或双引号 `"..."` 包起来。它是**不可变(immutable)**的:一旦创建,里面的字符就不能改。

### 索引和切片

- **索引**:`s[0]` 是第一个字符,`s[-1]` 是最后一个(负数从右往左数)
- **切片**:`s[起点:终点:步长]`,**左闭右开**(含起点、不含终点)
  - `s[0:3]` → 第 0、1、2 个字符
  - `s[::2]` → 每隔一个取
  - `s[::-1]` → 倒序
- 当我们定义 `a = "Python"` 时,计算机的执行顺序:
   - 先在内存中**创建字符串对象** `"Python"`(分配内存,自带 `id` / `type` / `value`)
   - 再创建变量名 `a`
   - 最后让 `a` **引用(指向)** 这个对象，即把对象的内存地址赋给a

  所以 `a` 不是"容器",只是指向对象的一个”指针“;两个变量可以指向同一个对象:

  ```python
  a = "Python"
  b = a                  # b 和 a 指向同一个对象
  print(id(a) == id(b))  # True,地址相同
  ```

### 常用方法

| 方法 | 作用 | 例子 | 结果 |
|---|---|---|---|
| `len(s)` | 长度 | `len("Miku")` | `4` |
| `s.upper()` | 全大写 | `"miku".upper()` | `"MIKU"` |
| `s.lower()` | 全小写 | `"MIKU".lower()` | `"miku"` |
| `s.strip()` | 去两端空白 | `" hi ".strip()` | `"hi"` |
| `s.replace(a, b)` | 替换 | `"a,b".replace(",", "-")` | `"a-b"` |
| `s.split(",")` | 按逗号拆分 | `"a,b".split(",")` | `["a", "b"]` |
| `s.find("b")` | 查找位置(找不到返回 -1) | `"abc".find("b")` | `1` |
| `"-".join(list)` | 用分隔符连接列表 | `"-".join(["a","b"])` | `"a-b"` |
| `"i" in s` | 判断是否包含 | `"i" in "Miku"` | `True` |

````python
s = "Hello, Miku"

# 索引和切片
print(s[0])        # H
print(s[-1])       # u
print(s[0:5])      # Hello
print(s[::2])      # Hlo iu (每隔一个)
print(s[::-1])     # ukiM ,olleH (倒序)

# 常用方法
print(len(s))                    # 11
print(s.upper())                 # HELLO, MIKU
print("  hi  ".strip())          # hi
print(s.replace("Miku", "world"))  # Hello, world
print(s.split(","))              # ['Hello', ' Miku']
print("i" in s)                  # True

# 字符串是不可变的:下面这行会报错,可以取消注释试试
# s[0] = "h"   # TypeError: 'str' object does not support item assignment
````

### 2.3 List 列表

列表用方括号 `[...]` 包起来,元素可以是任意类型。

和字符串**最大的不同:列表是可变(mutable)的**——能改元素、增删元素。索引和切片规则与字符串**完全一样**。

##### 常用方法

| 方法 | 作用 |
|---|---|
| `lst.append(x)` | 末尾添加一个元素 |
| `lst.insert(i, x)` | 在位置 i 插入 |
| `lst.pop()` | 删除并返回末尾元素 |
| `lst.remove(x)` | 删除第一个值为 x 的元素 |
| `lst.sort()` | 原地排序(改原列表) |
| `sorted(lst)` | 返回新列表(不改原列表) |
| `lst.reverse()` | 原地反转 |
| `lst.index(x)` | 查找第一次出现的位置 |
| `x in lst` | 判断是否包含 |
| `len(lst)` | 长度 |

````python
lst = [1, 2, 3, 4, 5]
print(lst[0])      # 1
print(lst[-1])     # 5
print(lst[1:3])    # [2, 3] 左闭右开

# 可变:直接改元素
lst[0] = 99
print(lst)         # [99, 2, 3, 4, 5]

# 常用方法
nums = [3, 1, 2]
nums.append(4)
print(nums)        # [3, 1, 2, 4]
nums.sort()
print(nums)        # [1, 2, 3, 4]
print(nums.pop())  # 4 (弹出末尾)
````

##### 易错点

列表是可变对象,`b = a` 只是让 `b` 也指向**同一个列表**:

```python
a = [1, 2, 3]
b = a            # 同一个对象
b.append(4)
print(a)         # [1, 2, 3, 4]  ← a 也跟着变了!
```

想得到**独立副本**,用 `b = a.copy()` 或 `b = a[:]`。

##### 列表推导式

语法:`[表达式 for 变量 in 可迭代对象]`,可加 `if` 过滤:

```python
[x**2 for x in range(5)]              # [0, 1, 4, 9, 16]
[x for x in range(10) if x % 2 == 0]  # [0, 2, 4, 6, 8]
```

````python
# 引用陷阱
a = [1, 2, 3]
b = a              # 同一个对象
b.append(99)
print(a)           # [1, 2, 3, 99]  ← a 也变了!

c = a.copy()       # 独立副本
c.append(100)
print(a, c)        # [1, 2, 3, 99] [1, 2, 3, 99, 100]

# 列表推导式
print([x**2 for x in range(5)])              # [0, 1, 4, 9, 16]
print([x for x in range(10) if x % 2 == 0])  # [0, 2, 4, 6, 8]
````

#### 2.4 Tuple 元组

元组用圆括号 `(...)` 定义,和列表很像,但有一个关键区别:**元组是不可变(immutable)的**——创建后不能增删改元素。

| | 列表 list | 元组 tuple |
|---|---|---|
| 符号 | `[1, 2, 3]` | `(1, 2, 3)` |
| 可变性 | 可变 | **不可变** |
| 用途 | 需要增删改的数据 | 不想被改的数据(坐标、配置) |

**几个要点**

- 索引和切片规则与列表 / 字符串**完全一样**
- **单元素元组要加逗号**:`(1,)` 才是元组,`(1)` 只是数字 `1`
- **解包**:`x, y = (1, 2)` 一行给多个变量赋值
- 元组可以作为字典的 key,列表不行(因为元组不可变、可哈希)

````python
t = (1, 2, 3, 4, 5)
print(type(t))       # <class 'tuple'>
print(t[0])          # 1
print(t[1:3])        # (2, 3)

# 不可变:下面这行会报错
# t[0] = 99   # TypeError: 'tuple' object does not support item assignment

# 单元素元组要加逗号
print(type((1,)))    # <class 'tuple'>
print(type((1)))     # <class 'int'>  ← 这是数字 1,不是元组!

# 解包
x, y, z = (10, 20, 30)
print(x, y, z)       # 10 20 30

# 交换两个变量(利用解包)
a, b = 1, 2
a, b = b, a
print(a, b)          # 2 1
````

#### 2.5 Set 集合

集合用花括号 `{...}` 定义,有两个特点:**元素无序、不重复**。

**主要用途**

- **去重**:自动去掉重复元素
- **集合运算**:交集、并集、差集(就是数学里的集合)

**常用操作**

| 操作 | 含义 |
|---|---|
| `set([...])` | 从列表重建集合(去重) |
| `s.add(x)` | 添加元素 |
| `s.remove(x)` | 删除元素 |
| `a & b` | 交集(两边都有) |
| `a \| b` | 并集(任一边有) |
| `a - b` | 差集(在 a 不在 b) |
| `x in s` | 判断是否包含 |

> 集合不能索引(无序),也不能放可变元素(比如列表)。

````python
# 去重
nums = [1, 2, 2, 3, 3, 3]
print(set(nums))       # {1, 2, 3}

# 集合运算
a = {1, 2, 3}
b = {2, 3, 4}
print(a & b)           # {2, 3} 交集
print(a | b)           # {1, 2, 3, 4} 并集
print(a - b)           # {1} 差集

# 添加 / 删除
s = {1, 2}
s.add(3)
print(s)               # {1, 2, 3}
````

#### 2.6 Dict 字典

字典用 `{key: value}` 定义,存的是**键值对**,像查表一样通过 key 找 value。

**要点**

- **键 key 必须是不可变的**(字符串、数字、元组),值 value 可以是任意类型
- 取值:`d[key]`;但 key 不存在会报错,用 `d.get(key, 默认值)` 更安全
- 增改:`d[key] = value`(键存在就改,不存在就加)

**常用方法**

| 方法 | 作用 |
|---|---|
| `d[key]` | 取值(键不存在会报错) |
| `d.get(key, 默认)` | 安全取值(不存在返回默认值) |
| `d[key] = v` | 添加或修改 |
| `del d[key]` | 删除 |
| `key in d` | 判断键是否存在 |
| `d.keys()` | 所有键 |
| `d.values()` | 所有值 |
| `d.items()` | 所有键值对(遍历用) |

````python
d = {"name": "Miku", "age": 16}
print(d["name"])              # Miku
print(d.get("height", "unknown"))  # unknown (键不存在,返回默认)

# 添加 / 修改
d["age"] = 16                 # 修改
d["city"] = "Tokyo"           # 添加
print(d)                      # {'name': 'Miku', 'age': 16, 'city': 'Tokyo'}

# 遍历键值对
for k, v in d.items():
    print(k, "=", v)

# 判断键是否存在
print("age" in d)             # True
````

## 3. Python 三大程序结构

### 3.1 顺序结构

程序默认按**从上到下的顺序,一行一行执行**,这就是顺序结构——最基本、最自然的执行方式。

每条语句按书写顺序依次执行,执行完上面才轮到下面。

````python
print("Step 1")
print("Step 2")
print("Step 3")
# 依次输出:Step 1 → Step 2 → Step 3
````

### 3.2 分支结构

分支结构让程序根据**条件**走不同的路,用 `if` / `elif` / `else` 实现。

**基本语法**

```python
if 条件:
    满足条件时执行的代码
elif 另一个条件:
    满足另一个条件时执行的代码
else:
    都不满足时执行的代码
```

**三个要点**

- 条件后面要写**冒号 `:`**
- 冒号下面要**缩进 4 个空格**(Python 靠缩进区分代码块,不是大括号!)
- `elif`、`else` 都可以省略

**比较运算符**:`==` 等于、`!=` 不等于、`>` `<` `>=` `<=`

**逻辑运算符**:`and`(且)、`or`(或)、`not`(非)

````python
score = 85

if score >= 90:
    print("Excellent")
elif score >= 60:
    print("Pass")
else:
    print("Fail")

# 逻辑运算:and 表示"两个条件都满足"
age = 20
if age >= 18 and age < 60:
    print("adult")
````

### 3.3 循环结构

循环结构让程序重复执行一段代码,有两种:

### `for` 循环 —— 遍历一个序列

```python
for 变量 in 可迭代对象:
    循环体
```

`range(n)` 生成 `0, 1, 2, ..., n-1`(不含 n)。可以遍历列表、字符串、字典等。

### `while` 循环 —— 条件成立就一直跑

```python
while 条件:
    循环体
```

**要点**

- 和 if 一样:冒号 + 缩进
- `break` 直接跳出**整个循环**
- `continue` 跳过**本次**,进入下一次
- while 要小心**死循环**(条件一直为真,程序停不下来)

````python
# for + range:打印 0 到 4
for i in range(5):
    print(i, end=" ")   # 0 1 2 3 4
print()

# 遍历列表
for name in ["Miku", "Rin", "Len"]:
    print("Hello,", name)

# while:累加到 100
total = 0
i = 1
while i <= 100:
    total += i
    i += 1
print("Sum from 1 to 100:", total)   # 5050

# break 和 continue
for i in range(10):
    if i == 3:
        continue   # 跳过 3
    if i == 8:
        break      # 到 8 就停
    print(i, end=" ")
# 输出:0 1 2 4 5 6 7
````

````python
# 打印九九乘法表
## while 循环
i = 1
while(i < 10):
    j = 1;
    while(j <= i):
        print("%d*%d=%d" % (i,j,i*j),end = '\t')
        j = j + 1
    i = i + 1
    print("\n")

## for 循环
for i in range(1,10):
    for j in range(1,i+1):
        print(f"{i}*{j}={i*j}", end = '\t')
    print("\n")
````
